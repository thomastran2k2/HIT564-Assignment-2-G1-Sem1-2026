```python
import requests
from bs4 import BeautifulSoup
import os
from datetime import datetime
```

```python

BASE = "https://data.nt.gov.au"
ORG_URL = "https://data.nt.gov.au/organization/department-of-attorney-general-and-justice"
ALC_URL = "https://data.nt.gov.au/organization/tourism-and-hospitality"

```

```python
def get_datasets():
    """Scrape all datasets from the AGD organization page."""
    html = requests.get(ALC_URL).text
    soup = BeautifulSoup(html, "html.parser")

    datasets = []
    for item in soup.select(".dataset-item"):
        title = item.select_one(".dataset-heading a").text.strip()
        link = BASE + item.select_one(".dataset-heading a")["href"]
        datasets.append((title, link))
    return datasets
```

```python
def get_csv_link(dataset_url, name):
    """Open dataset page and extract the CSV resource link."""
    html = requests.get(dataset_url).text
    soup = BeautifulSoup(html, "html.parser")
    link = ""
    if name == "crime":
        link = soup.find("a", class_="resource-url-analytics")["href"]
    if name == "alcohol":
        items = soup.select("li.resource-item")

    for item in items:
        # Title
        title = item.select_one("a.heading").get_text(strip=True)

        # File format (CSV, XLSX, PDF)
        fmt = item.select_one(".format-label").get_text(strip=True)
        
        # Download link
        link = item.select_one("a.resource-url-analytics")["href"]
        if fmt =='xlsx' or fmt == 'csv':
            return link
        
    return link
```

```python

def download_csv(url, year, month):
    """Download CSV into structured folder."""
    folder = f"crime_stats/{year}"
    os.makedirs(folder, exist_ok=True)

    filename = f"{month:02d}.csv"
    path = os.path.join(folder, filename)

    print(f"Downloading {filename} ...")
    r = requests.get(url)
    r.raise_for_status()

    with open(path, "wb") as f:
        f.write(r.content)
```
